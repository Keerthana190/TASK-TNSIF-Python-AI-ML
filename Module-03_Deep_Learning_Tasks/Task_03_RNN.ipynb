##Practice Example 1: Temperature Prediction using RNN
Real-Life Application

Weather forecasting systems use RNNs to analyze previous temperature records and predict future temperatures.

```python
# ==========================================
# RNN Practice Example-1
# Temperature Prediction
# ==========================================

# Step 1: Import Libraries

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# ------------------------------------------
# Step 2: Create Dataset
# ------------------------------------------

# Previous 3 days temperature
X = np.array([
    [[30],[31],[32]],
    [[31],[32],[33]],
    [[32],[33],[34]],
    [[33],[34],[35]],
    [[34],[35],[36]]
])

# Next day's temperature
y = np.array([33,34,35,36,37])

# ------------------------------------------
# Step 3: Build RNN Model
# ------------------------------------------

model = Sequential([

    SimpleRNN(16, activation="tanh", input_shape=(3,1)),

    Dense(1)

])

# ------------------------------------------
# Step 4: Compile Model
# ------------------------------------------

model.compile(

    optimizer="adam",

    loss="mse"

)

# ------------------------------------------
# Step 5: Train Model
# ------------------------------------------

model.fit(

    X,

    y,

    epochs=200,

    verbose=0

)

print("Training Completed!")

# ------------------------------------------
# Step 6: Predict
# ------------------------------------------

test = np.array([[[35],[36],[37]]])

prediction = model.predict(test)

print("\nPredicted Temperature for Next Day:")

print(round(prediction[0][0],2))
```

###Practice Example 2: Monthly Sales Prediction using RNN
Real-Life Application

Retail companies use RNN models to forecast future sales based on historical monthly sales data. Accurate sales prediction helps businesses manage inventory, production, and marketing strategies.

```python
# ==========================================
# RNN Practice Example-2
# Monthly Sales Prediction
# ==========================================

# Step 1: Import Libraries

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# ------------------------------------------
# Step 2: Create Dataset
# ------------------------------------------

# Sales of previous 3 months
X = np.array([
    [[100],[120],[140]],
    [[120],[140],[160]],
    [[140],[160],[180]],
    [[160],[180],[200]],
    [[180],[200],[220]]
])

# Next month's sales
y = np.array([160,180,200,220,240])

# ------------------------------------------
# Step 3: Build Model
# ------------------------------------------

model = Sequential([

    SimpleRNN(

        units=16,

        activation="tanh",

        input_shape=(3,1)

    ),

    Dense(1)

])

# ------------------------------------------
# Step 4: Compile Model
# ------------------------------------------

model.compile(

    optimizer="adam",

    loss="mse"

)

# ------------------------------------------
# Step 5: Train Model
# ------------------------------------------

model.fit(

    X,

    y,

    epochs=200,

    verbose=0

)

print("Training Completed!")

# ------------------------------------------
# Step 6: Predict
# ------------------------------------------

test = np.array([[[200],[220],[240]]])

prediction = model.predict(test)

print("\nPredicted Next Month Sales:")

print(round(prediction[0][0],2))
```
