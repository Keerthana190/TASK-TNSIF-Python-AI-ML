# What is a Data Preprocessing Pipeline?

## Definition

A Data Preprocessing Pipeline is a sequence of data processing steps performed automatically before training a Machine Learning model.

Instead of manually cleaning, scaling, encoding, and training the model one by one, a pipeline performs all these steps together.

## Formula

Raw Data
     ↓
Preprocessing
     ↓
Model Training
     ↓
Prediction

## Real-Life Example

Suppose you are predicting house prices.

Before training the model, you need to:

- Fill missing values
- Scale numerical columns
- Encode categorical columns

A Pipeline performs all these tasks automatically whenever new data is given.

# Why do we use Pipelines?

Without Pipeline

- Clean Data
- Scale Data
- Train Model
- Predict

Every time we receive new data, we repeat these steps manually.

With Pipeline

All preprocessing and model training happen automatically in one step.

This saves time and reduces errors.

# Pipeline Workflow

Collect Data
      ↓
Handle Missing Values
      ↓
Feature Scaling
      ↓
Encoding
      ↓
Train Model
      ↓
Prediction

```python
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.DataFrame({

    "Age":[22,25,47,52,46,56,48,60],

    "Salary":[25000,30000,65000,72000,62000,80000,75000,90000],

    "Purchased":[0,0,1,1,1,1,1,1]

})

df
```

```python
X = df[["Age","Salary"]]

y = df["Purchased"]
```

```python
pipeline = Pipeline([

    ("scaler", StandardScaler()),

    ("model", LogisticRegression())

])

pipeline.fit(X,y)
```

```python
prediction = pipeline.predict([[35,50000]])

print("Prediction:", prediction[0])
```

# What is ColumnTransformer?

## Definition

ColumnTransformer allows different preprocessing techniques to be applied to different columns in the dataset.

Example:

- Scale numerical columns
- Encode categorical columns

It combines all preprocessing steps into one object.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({

    "Age":[22,25,47,52],

    "City":["Hyderabad","Warangal","Hyderabad","Karimnagar"]

})

df
```

```python
transformer = ColumnTransformer(

    transformers=[

        ("num", StandardScaler(), ["Age"]),

        ("cat", OneHotEncoder(), ["City"])

    ]

)

result = transformer.fit_transform(df)

print(result)
```
